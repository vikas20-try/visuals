```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("yufengsui/mobile-games-ab-testing")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

#reading the data
#joined_path = os.path.join(path,"ab_testing.csv")
df0 = pd.read_csv("cookie_cats.csv")
df = df0.copy()
df.head()
```

```python
#finding bot making more rounds and only retrain
df = df[df.sum_gamerounds < 20000]
len(df)
df.head()
```

```python
#here we are testing the retention btw gate 30 and gate 40 version
# let us assume the gate 30 is A and gate 40 is B
ret1 = df.groupby('version')['retention_1'].mean()
ret7 = df.groupby('version')['retention_7'].mean()
print(ret1)
print(ret7)
```

```python
print("conclusion was gate 30 in winner here")
```

```python
#since the rention is not a number but a boolean value we can't use the t-test
#so we are using the chi-sqaures test to get the stats
from scipy.stats import chi2_contingency
#making a cross tab for the test
contingency = pd.crosstab(df['version'],df['retention_7'])
chi,p_value,doff,expected = chi2_contingency(contingency)
print(f"7 days value is {p_value:.4f}")
```

```python
# initial conclusion that there is no signficant diff btw these two version
# drawing a plot
contingency.plot(kind='bar',stacked=True)
plt.show()
```

```python
boot = []
for i in range(1000):
  boot_mean = df.sample(frac = 1,replace=True).groupby('version')['retention_7'].mean()
  boot.append(boot_mean)
#converting it into the dataframe
df_plot = pd.DataFrame(boot)
df_plot.plot(kind ='kde',title="Bootstarp 7 days retnetion")
plt.show()
```

```python
import seaborn as sns
plt.figure(figsize=(8, 5))
# If boot_1d is stored in a long-format DataFrame
sns.violinplot(data=boot, palette="muted")

plt.title("Spread of Bootstrapped Means by Game Version")
plt.ylabel("Retention 1-Day Mean")
plt.xlabel("Version")
plt.show()
```
